```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

data = fetch_california_housing(as_frame=True)
df = data.frame

# Filter houses with average rooms > 5
print(df[df["AveRooms"] > 5])
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
df = data.frame

# Multiple conditions
filtered = df[(df["AveRooms"] > 5) & (df["MedInc"] > 3)]

print(filtered)
```

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

data = fetch_california_housing(as_frame=True)
df = data.frame

# Create missing values
df.loc[0:4, "MedInc"] = None

# Fill missing values
df["MedInc"] = df["MedInc"].fillna(df["MedInc"].mean())

print(df.head())
```

```python
from sklearn.datasets import fetch_california_housing
import pandas as pd

data = fetch_california_housing(as_frame=True)
df = data.frame

# Find missing values
print(df.isnull().sum())
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
df = data.frame

# Remove rows with missing values
df = df.dropna()

print(df)
```

```python
from sklearn.datasets import fetch_california_housing

data = fetch_california_housing(as_frame=True)
df = data.frame

# Remove duplicate rows
df = df.drop_duplicates()

print(df)
```

```python
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt

data = fetch_california_housing(as_frame=True)
df = data.frame

# Scatter plot
plt.scatter(df["MedInc"], df["MedHouseVal"])
plt.xlabel("Median Income")
plt.ylabel("House Value")
plt.show()
```

```python
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt

data = fetch_california_housing(as_frame=True)
df = data.frame

# Pie chart using first 5 house values
df.head(5).set_index("MedInc")["MedHouseVal"].plot(
    kind="pie", autopct="%1.1f%%"
)
plt.ylabel("")
plt.show()
```

```python
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt

data = fetch_california_housing(as_frame=True)
df = data.frame

plt.subplot(1, 2, 1)
plt.plot(df["MedInc"].head(10))
plt.title("Median Income")
plt.xlabel("Index")
plt.ylabel("Income")

plt.subplot(1, 2, 2)
plt.plot(df["MedHouseVal"].head(10))
plt.title("House Value")
plt.xlabel("Index")
plt.ylabel("Value")

plt.tight_layout()
plt.show()
```
